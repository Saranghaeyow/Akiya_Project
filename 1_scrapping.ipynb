```python
import pandas as pd
from bs4 import BeautifulSoup
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.support.ui import Select
import time
```

```python

website = 'https://www.akiyajapan.com/search/list?categories%5B%5D=buy&sizeUnit=sqm&categories%5B%5D=house&order_by=date_desc'
#path = r'D:\programming files\Python\Jupyter\chromedriver-win64\chromedriver.exe' #chrome driver file path
#service = Service(path)
#driver = webdriver.Chrome(service=service)
driver = webdriver.Chrome() 
driver.get(website)

table_view = driver.find_element(By.XPATH, '//*[@id="table-view-btn"]') #this line is about calling button for it to click auto 
table_view.click()


#drop_down = Select(driver.find_element(By.ID, 'country'))
#drop_down.select_by_index(1)

time.sleep(3)

soup = BeautifulSoup(driver.page_source, 'html.parser')
```

```python
table = soup.find_all('tr')
```

```python
for match in table:
    tds = match.find_all('td')

    cleaned_cells = []

    for i, td in enumerate(tds):
        # 🖼️ Handle image separately (first column)
        if i == 0:
            img_tag = td.find('img')
            if img_tag:
                image = img_tag.get('src') or img_tag.get('data-src') or ''
            else:
                image = ''
            cleaned_cells.append(image)
        else:
            text = td.text.strip()
            parts = text.split('\n')

            for part in parts:
                if part.strip():
                    cleaned_cells.append(part.strip())

    # Ensure fixed length
    while len(cleaned_cells) < 7:
        cleaned_cells.append('')

    image, location, price, types, beds, house, konbini = cleaned_cells[:7]

    print([image, location, price, types, beds, house, konbini])
```

```python

last_location = ''

for match in table:
    tds = match.find_all('td')

    if len(tds) >= 5:
        image = tds[0].text.strip()
        
        location = tds[1].text.strip()
        if not location:
            location = last_location
        else:
            last_location = location

        price = tds[2].text.strip()
        types = tds[3].text.strip()
        beds = tds[4].text.strip()
        house = tds[5].text.strip() if len(tds) > 5 else ''

        print([image, location, price, types, beds, house])
    #length = len(df)
    #df.loc[length] = [date, home, score, away]      
```

```python
akiya = []

for tables in table[1:]:
    tds = tables.find_all('td')

    # 🖼️ Image
    img_tag = tds[0].find('img') if len(tds) > 0 else None
    image = img_tag.get('src') or img_tag.get('data-src') if img_tag else ''
    price = tds[2].text.strip() if len(tds) > 2 else ''
    type = tds[3].text.strip() if len(tds) > 3 else ''
    beds = tds[4].text.strip() if len(tds) > 4 else ''
    land = tds[5].text.strip() if len(tds) > 5 else ''
    house = tds[6].text.strip() if len(tds) > 6 else ''
    konbini = tds[7].text.strip() if len(tds) > 7 else ''

    # 📍 Location (STRICT: only accept real location)
    if len(tds) > 1: # this condition will check if there is  a value and if not it will return ''
        raw = tds[1].text.strip()

        # ✅ Only accept if it's really a location
        if raw and ',' in raw: # this code will check if there is a comma in the location which is a common format for locations (e.g., "Tokyo, Japan")
            location = raw.split('\n')[0]
        else:
            location = ''   # 👈 ALWAYS empty if invalid
    else:
        location = ''

    akiya.append([image, location, price, type, beds, land, house, konbini])

    print([image, location, price, type, beds, land, house, konbini])
```
